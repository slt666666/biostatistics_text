**Q. 課題２をヒストグラムで表現して調べる意義が実感できなかったのでこの表現方法が有利になる視点を教えていただきたいです。**

A. 今回のデータの場合、バイオリンプロットや種・島を同時に表した箱ひげ図など、他のグラフでも確認はできます。ただし、箱ひげ図は分布を四分位などに要約した図なので、二峰型の分布（今回でいうとisland別に描いた場合のBiscoe島）の形は表現できません。

種・島を組み合わせた箱ひげ図が描けるのは、「種」という要因が効いていそうだとあらかじめわかっているからです。もし種の情報がデータになかったとしたら、島別の箱ひげ図では「Biscoe島はばらつきが大きい」としか読み取れませんが、ヒストグラムであれば山が2つあることから「性質の異なる集団が混ざっているのでは」と気づくことができます。どの要因が重要かわからない探索の段階で、分布の形そのものを確認できるのがヒストグラム(/バイオリンプロット)の利点です。

````r
df <- read.csv("https://raw.githubusercontent.com/slt666666/biostatistics_text/refs/heads/main/source/_static/data/chapter3_penguins.csv")

library(ggplot2)
g <- ggplot(data=df, aes(x=flipper_length_mm))
g <- g + geom_histogram(aes(fill=island))
g

g <- ggplot(data=df, aes(x=island, y=flipper_length_mm))
g <- g + geom_boxplot()
g <- g + geom_jitter()
g
````

**Q. 自由度がなぜn-1なのか、何を意味しているかをもう一度教えていただきたいです。**

A. **自由度が何を意味しているか**

自由度は簡単に言うと「自由に決められる値の数」を表しています。

一番わかりやすい例で言うと平均値$\bar{X}$が決まっているときに、$X_1, X_2, ... X_{n-1}$の値が決まると、$X_n$は自動的に決まります。

例) A,B,Cさんがテストを受け、平均点が60点だった。Aさんは50点、Bさんは60点だった時、Cさんの点数は自動的に70点と決まります。

<br>

これを分散の計算で考えると、分散は、各データが平均からどれだけずれているか（偏差）をもとに計算します。ところが、偏差は合計すると必ず0になります。

そのため、n個の偏差のうち n−1個がわかれば、最後の1個は自動的に決まってしまいます。ばらつきについての独立した情報は、n個ではなく n−1個分しかないということです。

先ほどのテストの例で言うと、3人の点数のばらつきは、「AさんとBさんの差（10点）」と「BさんとCさんの差（10点）」の2つがわかれば、「AさんとCさんの差（20点）」は、10 + 10 で計算できます。つまり、3つの偏差のうち自由に決まるのは2つだけで、自由度は 3 − 1 = 2 です。

(これらの計算では自由度は$n-1$ですが、別の計算では自由度が更に小さくなるようなものもあります。)

**不偏分散をn-1で割る理由**

$s^2=\dfrac{1}{n-1}\lbrace(X_1-\bar{X})^2+(X_2-\bar{X})^2+...+(X_n-\bar{X})^2\rbrace$

ここで、標本分散$s^2$についても、期待値$E(s^2)=\sigma^2$になるから、自由度は$n-1$で割るという説明しかしませんでした、

理論的に説明するにはちょっと時間を要するので、直感的に次のように説明できます。

標本平均は、そのデータ自身のちょうど真ん中にくるように計算されます。そのため、データは「本当の平均」よりも「標本平均」のほうに近く見えます。

その結果、標本平均からの偏差で計算したばらつきは、本当のばらつきより小さめに出てしまいます。それを補正するために、$n$ より少し小さい $n−1$ で割ります。

これは、R のシミュレーションで実際に見ると納得しやすくなります。

母平均50, 母分散100の母集団から5個体とって分散を計算する、というのを10000回くりかえします。

このとき$n$ で割った場合と、 $n−1$ で割った場合どうなるかを比較してみます。

````r
set.seed(1)
n <- 5   # 1回に取るデータの数
## 本当の平均50・本当の分散100（標準偏差10）の母集団から、5個取る実験を1万回くり返す
result <- replicate(10000, {
  x <- rnorm(n, mean = 50, sd = 10)
  c("nで割る"                 = sum((x - mean(x))^2) / n,
    "n-1で割る"               = sum((x - mean(x))^2) / (n - 1),
    "本当の平均を使ってnで割る" = sum((x - 50)^2) / n)
})
round(rowMeans(result), 1)   # 1万回の平均（本当の分散は100）
````

$n$で割った場合、母分散より小さく出ることが多いことが分かります。

本当の平均がわかっていれば、$n$ で割っても大丈夫そうなので、小さく出た原因が標本平均を使ったことにあると確かめられます。
