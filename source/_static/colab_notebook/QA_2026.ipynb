# Q & A

# 第2回の質問



**Q. 出席点はありますか。**

A. シラバス見てください。

<br>

**Q. Google Colabを使い始めたばかりの初心者なのですが、このツールは多くのリソースを消費するのでしょうか？Colabを実行すると、PCの動作がかなり重くなってしまうようです。PCの性能があまり高くない場合、RStudioを使ったほうがよいのでしょうか？。**

A. Google Colab自体はリソースは消費しません(Googleのサーバー上で動いているので、自分のPCのリソースは使用しない)。PCが重くなる原因はブラウザ側(Google Chromeとか)にあるように思います。ブラウザの再起動など試してみてください。

<br>

**Q. 授業内で扱ったサッカーのデータが上手く読み込まれませんでした。このようなときはAIに聞けばよいのでしょうか。**
**Q. 友達と同じように入力していてもエラーになることがたまにあるのですがどうしてそうなりますか？**

A. まずエラー文を読むと原因が分かることが多いです。AIに聞く場合もエラー文をコピペして聞くと原因が分かる可能性が高いです。ファイル名等の間違いが一番多い気がします。

<br>

**Q. 課題で最初に表示されていたデータフレームを実行すると、表で６人分のデータが表示されていたのに対し、課題１,3, 4をするときには授業中に読み込んだ大量のデータ(多分100人以上）が用いられていたように思いますが、読み込むデータが間違っていたのでしょうか？**

A. 課題の最初で使用していた`head(データフレーム変数)`という関数はデータフレームの最初の方だけを見せる機能を持っています。

<br>

**Q. 課題②と③の答えがどうやっても同じになります。なぜでしょうか。**

A. 同じになるのが正しいです。あえて意味のない条件付けをさせています。

<br>

**Q. 課題をどのような形式で提出すればいいのか明らかにしてほしいです。**

A. シンプルに答えを記述してもらえればOKです。コードとかはあっても無くてもどちらでも可です。

<br>

**Q. 課題を実行した後、出力をコピーしようとしたところ、数字だけがコピーされ、数値と数値を区切る「・」はコピーされませんでした。「・」もテキストとして認識させるための方法は何かあるのでしょうか。**

**Q. 出した数値の点がコピペの時に消えてしまい、数字がつながってしまうのですがその解決方法はありますか。**

A. ちょっとこちらで再現できないのでOSやブラウザ何を使ってそうなったか教えてもらえたら助かります！

<br>

**Q. 	プログラミングを書く際、演算子の両側にスペースを空けたほうがよいのか知りたい。**

A. 好みのところもあります。`1+1`が好きな方もいれば`1 + 1`を好む方もいます。(どちらも正しく動きます。)

<br>

**Q. 授業のファイル中に**`df[(df$position == "Defender") & (df$age <= 20), 'goals_overall']`**とあります。”（ダブルクオーテーション）と’（シングルクオーテーション）には意味的な違いがあるのでしょうか。堺先生の中で区別している点などがあれば知りたいです。**

A. RやPythonではどちらでも同じ意味なので、片方に統一していれば良いと思います。個人的にはダブルが好みです。他のプログラミング言語ではそれぞれ違った意味を持っていたりします。(テキストはたまに急いで修正してるところとかあるのでバラバラです…。)

<br>

**Q. Google Colab でR言語を使いたいと思ったとき、どのように設定すればR言語が使えるのか教えていただきたいです。実習用のものではなく、自分でGoogle Colabを開いたところ、2^10の実行結果が8になってしまいました。**

A. デフォルトだとGoogle ColabではPythonが動きます。上のメニューから「ランタイム」→「ランタイムのタイプを変更」から切り替えられます。
