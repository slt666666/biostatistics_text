**Q. 分散について、手で記述した分散の出し方と不偏分散の違いがよく理解できなかった。母数に依存することはわかったがうまく利用できず課題は不偏分散で計算した。また課題の４つ目で種ごとに分析する方法がわからずAIに頼ったところtapply関数が登場したので、来週の授業が楽しみである。**

A. 不偏分散については次回か次々回の講義で扱う予定です。基本的に統計の世界では不偏分散を用います。

`tapply`関数はデータをある因子毎(今回だとペンギンの種類ごと、とか)に処理するときに用いる関数です。

```r
df <- read.csv("https://raw.githubusercontent.com/slt666666/biostatistics_text_wed/refs/heads/main/source/_static/data/chapter3_penguins.csv")

head(df)
```

```r
# dfのspeciesの要素毎にbill_length_mmの平均値を算出
tapply(df$bill_length_mm, df$species, mean)
```

**Q. コードを書くときに、一気に複数のコードを実行してしまうと、どのコードとどの数値が対応しているのかとても分かりにくくなってしまうのですが、どのようにすれば分かりやすく書けるのでしょうか。**

A. 複数の値が出力されて分かりづらい場合、print関数を使用して説明なども一緒に出力すると良いです。


コードの内部でどの変数に何が入ってるか分かりづらい場合は変数名を意味のあるものにしておくと良いです。

<br>

```r
print("平均値")
print(mean(df$bill_length_mm))

print("中央値")
print(median(df$bill_length_mm))

# paste関数で文字列をつなげることも可
print(paste("標準偏差", sd(df$bill_length_mm)))
```
